----

Создать базовый класс Subscription в C#, который будет представлять подписки на
различные услуги. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

```
public interface ISubscription
{
    int SubscriptionId{ get; }
    string ServiceName{ get; }
    decimal CalculateMonthlyCost(); 
    void ExtendSubscription(int months);
    string GetSubscriptionDetails();
}

public interface IExtendableSubscription : ISubscription 
{
    void ExtendWithOffer(int months);
}

public abstract class SubscriptionBase : ISubscription
{
    public int SubscriptionId{ get;}
    public string ServiceName{ get; }
    protected decimal Cost { get; }
    protected SubscriptionBase(int id , string name, decimal cost)
    {
        SubscriptionId = id;
        ServiceName = name; 
        Cost = cost; 
    }
    public virtual decimal CalculateMonthlyCost() => Cost;
    public virtual void ExtendSubscription(int months) => Console.WriteLine($"[{ServiceName}] продлена на {months} мес.");
    public virtual string GetSubscriptionDetails() => $"ID:{SubscriptionId}, Услуга {ServiceName}, Стоимость {Cost} ";
}
public class OnlineServiceSubscription : SubscriptionBase
{
    public int MaxUsers {get;}
    public OnlineServiceSubscription(int id, string name, decimal cost, int users)
        : base(id,name,cost) => MaxUsers = users;
    public override decimal CalculateMonthlyCost() => Cost + (MaxUsers- 1 ) * 100m;
    public override string GetSubscriptionDetails() => base.GetSubscriptionDetails() + $", Пользователей {MaxUsers}"; 
}
public class StreamingSubscription : SubscriptionBase, IExtendableSubscription 
{
    public int MaxStreams {get;}
    public StreamingSubscription(int id, string name, decimal cost, int streams): base(id,name,cost) => MaxStreams = streams;
    public override void ExtendSubscription(int months) => Console.WriteLine($"{ServiceName} продлена на {months} мес.");
    public void ExtendWithOffer(int months) => Console.WriteLine($"{ServiceName} продлена на {months}  мес. +1 месяц бесплатно!");
    public override string GetSubscriptionDetails() => base.GetSubscriptionDetails() + $", Потоков:{MaxStreams}";
}
public class VideoSubscription : SubscriptionBase
{
    public string VideoQuality{get;}
    public VideoSubscription(int id, string name, decimal cost, string quality): base(id,name,cost) => VideoQuality = quality;
    public override string GetSubscriptionDetails() => base.GetSubscriptionDetails() + $", Качество:{VideoQuality}";
}
public class SubscriptionManager 
{
    private readonly ISubscription _subscription;
    public SubscriptionManager(ISubscription subscription) => _subscription = subscription;
    public void ShowInfo()
    {
        Console.WriteLine(_subscription.GetSubscriptionDetails());
        Console.WriteLine($"Стоимость в месяц: {_subscription.CalculateMonthlyCost()}");
    }
}
ISubscription[] subs = 
{
    new OnlineServiceSubscription(1,"OnlineCRM", 500m, 5 ),
    new StreamingSubscription(2,"StreamFlix", 700m, 3),
    new VideoSubscription(3,"VideoHub",300m, "4K HDR")
};
foreach (var s in subs)
{
    new SubscriptionManager(s).ShowInfo();
    s.ExtendSubscription(6);

    if (s is IExtendableSubscription ext)
        ext.ExtendWithOffer(6);
}
```
