Найти минимальное и максимальное значения в массиве.

```polyglot-notebook
int[] numbers = { 5, 8, 1, 3, 9, 6, 7, 2};

int min = numbers.Min();
int max = numbers.Max();
Console.WriteLine($"Min: {min}");
Console.WriteLine($"Max:{max}");
```

Найти сумму всех элементов массива.

```polyglot-notebook
int sum = numbers.Sum();
Console.WriteLine($"Sum: {sum}");
```

Найти среднее арифметическое всех элементов массива.

```polyglot-notebook
double average = numbers.Average();
Console.WriteLine($"среднее арифметическое: {average}");
```

Отсортировать массив в порядке возрастания

```polyglot-notebook
int[] sorted = numbers.OrderBy(x => x).ToArray();
Console.WriteLine($"Отсортированный массив: [{string.Join(",", sorted)}]");
```

Проверить, содержит ли массив определенное число

```polyglot-notebook
int target = 6; 
bool contains = numbers.Contains(target);
Console.WriteLine($"Массив содержит {target}: {contains}");
```
